```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../00_data/04_output/SAMIs_pop.csv")
df.loc[df['factor'] == 'gdp', 'factor'] = 'GDP'
df.loc[df['factor'] == 'mental_health_providers', 'factor'] = 'MHP'
df.loc[df['factor'] == 'DEPRESSION', 'factor'] = 'DEP'
df.loc[df['factor'] == 'children_single_parent', 'factor'] = 'SP'
df.loc[df['factor'] == 'BINGE', 'factor'] = 'ED'
df.loc[df['factor'] == 'pedestrian_3leg_intersections_total', 'factor'] = 'INT'
df.loc[df['factor'] == 'life_total', 'factor'] = 'LE'
df.loc[df['factor'] == 'DISABILITY', 'factor'] = 'DIS'
df.loc[df['factor'] == 'noise90n', 'factor'] = 'NOI'

df_fsami = pd.read_csv('../00_data/04_output/FsAMIs_pop_bestk.csv')
```

```python
df_fsami
```

```python
factors = ["DEP", "ED", "DIS", "SP", "MHP", "GDP", "INT"]

df_sami_fuc = (
    df[['factor','beta','p_value','CI_lower','CI_upper']]
    .drop_duplicates()
    .loc[lambda x: x['factor'].isin(factors)]
    .reset_index(drop=True)
)
df_fsami_fuc = (
    df_fsami[['factor','beta','p_value','CI_lower','CI_upper']]
    .drop_duplicates()
    .loc[lambda x: x['factor'].isin(factors)]
    .reset_index(drop=True)
)
```
